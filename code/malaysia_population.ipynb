```python
data["ethnicity"].unique()  
```

### Answer

Ethnicity in Malaysia in this dataset contains data in structure like this:

- **"overall"**: All ethnicity
- **"bumi_malay"**: Bumiputera Malay or Malaysian Malay 
- **"chinese"**: Malaysian Chinese
- **"indian"**: Malaysian Indian
- **"bumi_other"**: Other Malaysian Bumiputera Ethnicity
- **"other_citizen"**: Other Ethnicity that considered as a Malaysian Citizen.
- **"other_noncitizen"**: Other Ethnicity that not considered as a Malaysian Citizen.

# Data Preparation & Cleansing  

```python
print(data)  
```

```python
data["state"] = data["state"].replace("Pulau Pinang", "Penang", regex=False)

years = ["2020", "2021", "2022", "2023", "2024"]

for year in years:
    data["date"] = data["date"].replace(f"{year}-01-01", year, regex=False)  

clean_csv_data_file_path = output_path / "malaysia_population_by_district_clean.csv"
data.to_csv(clean_csv_data_file_path, index=False, encoding="utf-8")      
```

```python
data["population"] = data["population"] * 1000  
data["population"] = data["population"].fillna(0).astype("int64")   
data.to_csv(clean_csv_data_file_path, index=False, encoding="utf-8")    
```

## Sorting where in Malaysia that has Chinese the most.  

```python
# ethnicity = "chinese"      
states = ["Johor", "Kedah", "Kelantan", "Melaka", "Negeri Sembilan", "Pahang", "Perak", "Perlis", "Penang", "Sabah", "Sarawak", "Selangor", "Terangganu", "W.P. Kuala Lumpur", "W.P. Labuan", "W.P. Putrajaya"]
years = ["2020", "2021", "2022", "2023", "2024"]

for year in years:
  for state in states:
    output_data = data.loc[(data["state"] == state) & (data["sex"] == "both") & (data["age"] == "overall") & (data["date"] == year)]        
    output_data = output_data.drop(labels=["sex", "age"], axis=1, inplace=False)  
    output_data = output_data[output_data["ethnicity"] != "overall"]  
    output_data.sort_values(by=["ethnicity", "population"], ascending=False).to_csv(output_path / year / f"malaysian_{state}_population_{year}.csv", index=False, encoding="utf-8")    
```
